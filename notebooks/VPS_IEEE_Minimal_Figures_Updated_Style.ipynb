```python
def finish_figure(fig, filename):
    """Save a manuscript figure in raster and vector formats."""
    for ext in ("png", "pdf", "svg"):
        kwargs = {"dpi": 200} if ext == "png" else {}
        fig.savefig(OUTPUT_DIR / f"{filename}.{ext}", bbox_inches="tight", **kwargs)
    plt.show()
    plt.close(fig)


def grouped_age_bars(metric, ylabel, filename, ylim=None):
    methods = ["VPS-PPO", "VPS-G", "ValC", "RH2"]
    x = np.arange(len(subgroup_df))
    width = 0.19
    fig, ax = plt.subplots(figsize=(20, 18))
    for i, method in enumerate(methods):
        values = subgroup_df[f"{method}_{metric}"]
        ax.bar(
            x + (i - 1.5) * width,
            values,
            width,
            label=method,
            color=COLORS[method],
            edgecolor="black",
            linewidth=2.0,
            hatch=HATCHES[method],
        )
    ax.set_xticks(x, subgroup_df["subgroup"])
    ax.set_ylabel(ylabel)
    ax.set_ylim(bottom=0, top=ylim)
    ax.set_title("Conditioned Mean Age" if metric == "mean_age" else "Conditioned P95 Age", pad=24)
    ax.grid(axis="y", linestyle="--", alpha=0.5, linewidth=2.0)
    ax.set_axisbelow(True)
    ax.tick_params(axis="both", which="major", length=12, width=3)
    ax.legend(ncol=2, frameon=True, loc="upper left", handlelength=1.2,
              handleheight=0.8, handletextpad=0.4, borderpad=0.35, labelspacing=0.25)
    fig.tight_layout(pad=1.2)
    finish_figure(fig, filename)
```

## Concise interpretation

- The two subgroup plots answer only whether important states remain fresh. Statistical significance should be reported in the caption or text from the paired bootstrap analysis.
- The SUMO age and gain plots show that the RH2 target remained feasible but was repeatedly deferred despite its increasing immediate value.
- The BEV and two-track temporal plot demonstrate participant-specific policy differences on nuScenes without suggesting universal PPO dominance.
