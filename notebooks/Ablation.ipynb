```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Load the dataset from the provided table
data = {
    'Task Size': [
        12, 24, 36, 47, 60, 72, 84, 96, 108, 120, 132, 144, 156, 168, 180, 192,
    ],
    'FRODO': [
        0.9996, 0.9900, 0.9900, 0.9895, 0.9845, 0.9729, 0.9833, 0.9781,
        0.9876, 0.9750, 0.9801, 0.9806, 0.9837, 0.9693, 0.9619, 0.9519,
    ],
    'SFPD': [
        0.9833, 0.9781, 0.9876, 0.9750, 0.9671, 0.9598, 0.9650, 0.9615,
        0.9494, 0.9565, 0.9541, 0.9490, 0.9450, 0.9260, 0.9190, 0.9120,
    ],
    'PD': [
        0.9600, 0.9598, 0.9600, 0.9600, 0.9600, 0.9617, 0.9400, 0.9171,
        0.9500, 0.9420, 0.9400, 0.9300, 0.9150, 0.9300, 0.9150, 0.9110,
    ],
    'EPC': [
        0.9933, 0.9900, 0.9860, 0.9700, 0.9650, 0.9600, 0.9500, 0.9500,
        0.9410, 0.9400, 0.9300, 0.9100, 0.9000, 0.8800, 0.8750, 0.8600,
    ],
    'UC': [
        0.8600, 0.8800, 0.8900, 0.8600, 0.8700, 0.8400, 0.8300, 0.8200,
        0.8200, 0.8100, 0.8000, 0.8300, 0.8000, 0.7900, 0.7600, 0.7600,
    ],
    'SMDS': [
        0.9750, 0.9700, 0.9620, 0.9670, 0.9600, 0.9520, 0.9500, 0.9400,
        0.9360, 0.9310, 0.9320, 0.9200, 0.9150, 0.8700, 0.8600, 0.8300,
    ],
    'RU': [
        0.923140, 0.946408, 0.941210, 0.918450, 0.932150, 0.926830, 0.891240, 0.884500,
        0.902310, 0.879150, 0.871200, 0.865420, 0.878930, 0.846210, 0.818438, 0.873277,
    ],
}

df = pd.DataFrame(data).set_index('Task Size')

# Extract series
UC_avg_success = df['UC']
RU_avg_success = df['RU']
PD_avg_success = df['PD']
EPC_avg_success = df['EPC']
SFPD_avg_success = df['SFPD']
MDS_avg_success = df['SMDS']
PARE_avg_success = df['FRODO']

# --- Figure 1: Average SV Utilization ---
plt.figure(figsize=(20, 18))

UC_avg_success.plot(
    kind='line', color='limegreen', linewidth=10, marker='d', markersize=30, linestyle='-', label='UC'
)
RU_avg_success.plot(
    kind='line', color='darkviolet', linewidth=10, marker='<', markersize=30, linestyle='-', label='RU'
)
PD_avg_success.plot(
    kind='line', color='mediumslateblue', linewidth=10, marker='>', markersize=30, linestyle='-', label='PD'
)
EPC_avg_success.plot(
    kind='line', color='crimson', linewidth=10, marker='*', markersize=30, linestyle='-.', label='EPC'
)
SFPD_avg_success.plot(
    kind='line', color='orange', linewidth=10, marker='o', markersize=30, linestyle='--', label='SPC'
)
MDS_avg_success.plot(
    kind='line', color='black', linewidth=10, marker='^', markersize=30, linestyle='--', label='GMDS'
)
PARE_avg_success.plot(
    kind='line', color='blue', linewidth=10, marker='D', markersize=30, linestyle='--', label='FRODO'
)

plt.ylabel('Mean Success ratio', fontsize=80)
plt.xlabel(r'Average task dimension, $\overline{\omega}_{t}$', fontsize=80)

# Legend configuration
plt.legend(
    fontsize=60,
    loc='lower left',
    ncol=2,
    frameon=True,
    framealpha=0.8,
    facecolor='white',
    edgecolor='black',
    fancybox=True,
    shadow=False,
    columnspacing=1.0,
    handletextpad=0.5,
)

plt.xticks(fontsize=70)
plt.yticks(fontsize=70)
plt.grid(axis='both', alpha=1)
plt.tight_layout()
plt.ylim(0.6, 1.05)
plt.show()

xx = UC_avg_success.mean()
yy = RU_avg_success.mean()

print(f"UC mean: {xx}")
print(f"RU mean: {yy}")

# --- Figure 2: Relative Improvement vs. FRODO (Exact Match to Your Success Ratio Plot) ---
baselines = [
    (UC_avg_success,   'limegreen',       'd', '-',  'UC'),
    (RU_avg_success,   'darkviolet',      '<', '-',  'RU'),
    (PD_avg_success,   'mediumslateblue', '>', '-',  'PD'),
    (EPC_avg_success,  'crimson',         '*', '-.', 'EPC'),
    (SFPD_avg_success, 'orange',          'o', '--', 'SPC'),
    (MDS_avg_success,  'black',           '^', '--', 'GMDS'),
    (PARE_avg_success, 'blue',            'D', '--', 'FRODO'),
]

FRODO_vals = PARE_avg_success
improvements = {}
for series, _, _, _, label in baselines[:-1]:
    common_idx = series.index.intersection(FRODO_vals.index)
    if len(common_idx) == 0:
        imp = np.nan
    else:
        base_vals = series.reindex(common_idx)
        frodo_at_base = FRODO_vals.reindex(common_idx)
        imp = ((frodo_at_base - base_vals) / base_vals).mean() * 100
    improvements[label] = imp

plt.figure(figsize=(20, 18))

for i, (series, color, marker, linestyle, base_label) in enumerate(baselines):
    if base_label == 'FRODO':
        label = 'FRODO (ref)'
    else:
        imp = improvements[base_label]
        if np.isnan(imp):
            label = f'{base_label} (—%)'
        else:
            sign = '+' if imp >= 0 else ''
            label = f'{base_label} ({sign}{imp:.1f}%)'

    series.plot(
        kind='line',
        color=color,
        linewidth=10,
        marker=marker,
        markersize=30,
        linestyle=linestyle,
        label=label,
    )

plt.xlabel(r'Average task dimension, $\overline{\omega}_{t}$', fontsize=80)
plt.ylabel('Success ratio', fontsize=80)
plt.xticks(fontsize=60)
plt.yticks(np.arange(0.75, 1.01, 0.05), fontsize=60)
plt.ylim(0.75, 1.01)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.legend(
    loc='upper left',
    bbox_to_anchor=(0.02, 0.99),
    fontsize=55,
    frameon=True,
    fancybox=False,
    ncol=1,
    handlelength=0.8,
    handletextpad=0.2,
    borderpad=0.2,
    labelspacing=0.2,
)

plt.subplots_adjust(left=0.12, bottom=0.08, right=0.95, top=0.94)
plt.show()

# Print improvement table
print("\nFRODO Success Improvement (mean over task sizes):")
print("-" * 45)
for method, imp in improvements.items():
    if np.isnan(imp):
        print(f"{method:<6}: —")
    else:
        sign = '+' if imp >= 0 else ''
        print(f"{method:<6}: {sign}{imp:6.2f}%")
```
